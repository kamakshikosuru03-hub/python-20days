```python
"""guessing number program"""
guess = int(input("enter a number from 1 to 10:"))
while True:
  if guess> 10:
    print("invaild guess")
    break
  finder=int(input())
  if guess == finder:
    print("guessing number is unlocked!")
    break
  elif guess< finder:
    print("guessing value is too long try another")
  elif guess>finder:
    print("guessing value is small try another")
  else:
    print("invaild")
```
