```python
#leap year
year=int(input())
if (year%400==0):
  print("leap year")
elif (year%4==0 and year%100!=0):
  print("leap year")
else:
  print("not a leap year")
```

```python
#write prgm for gmail login
#username
#password
username=str(input())
password=str(input())
u=input()
p=input()
if username==u and password==p:
  print("login completed")
elif username==u and password!=p:
  print("password is incorrect")
elif username!=u and password==p:
  print("invalid username")
elif username!=u and password!=p:
  print("both login are failed")
else:
  print("login fail")
```

```python
#marks
Maths,Science,English=map(int,(input()).split())
sum=Maths+Science+English
avg=sum/3
print(sum)
print(avg)
if Maths<35 or Science<35 or English<35:
  print("F")
elif avg>=90 and avg<100:
  print("a")
elif avg>75 and avg<89:
  print("b")
elif avg>60 and avg<74:
  print("c")
elif avg>40 and avg<59:
  print("d")
else:
  print("f")
```
