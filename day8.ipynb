```python
#calucultor
a=int(input())
b=int(input())
for i in range(1,6):
  operator=input()
  if operator=='+':
        print(a+b)
  elif operator=='-':
       print(a-b)
  elif operator=='*':
      print(a*b)
  elif operator=='/':
       print(a/b)
  elif operator=='**':
       print(a**b)
  elif operator=='//':
       print(a//b)
  else:
      print("not a valid operator")
```
